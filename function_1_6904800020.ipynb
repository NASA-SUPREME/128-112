```python
#จงเขียนฟังก์ชันสามตัว
def is_valid_score(score):
  return score >= 0 and score <= 100

#แปลงคะแนนเป็นเกรด A (ตั้งเเต่ 80) B (ตั้งเเต่ 70) C (ตั้งเเต่ 60) และ F
def to_grade(score):
  if score >= 80:
    return"GRADE A"
  elif score >= 70:
    return"GRADE B"
  elif score >= 60:
    return"GRADE C"
  else:
    return"GRADE F"

#ทีแสดงผลในรูปแบบรายงาน โดยถ้าคะแนนไม่ถูกต้องให้แสดงข้อความแจ้งเตือนแทนเกรด ทุกฟังก์ชันต้องมี docstring
def print_result(student_id, score):
  if is_valid_score(score):
    print(f"รหัสนักเรียน {student_id}")
    print(f"ได้คะแนน {score} ได้เกรด {to_grade(score)}")
  else:
    print(f"คะแนน {score} ไม่ถูกต้อง")

print_result("6904800001",80)
print_result("6904800002",70)
print_result("6904800003",60)
```
