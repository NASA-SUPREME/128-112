```python
#โดยให้ทำการคำนวณหาราคารวมของรหัสคำสั่งซื้อแต่ละรายการ โดยให้คำนวณ vat 7% และราคารวม vat สุทธิ เก็บข้อมูล รหัสคำสั่งซื้ (id) ราคารวม และราคารวม vat ลงใน list ของ tuple จากนั้นแสดงรายงานยอดขายทุก tuple และยอดขายรวมทั้งหมด
orders = [
    {"id": 2001,"customer": "Anan", "city": "Korat","product":"Keyboard", "qty": 2, "price": 450},

    {"id": 2002,"customer": "Benja", "city": "Bangkok","product":"Mouse",    "qty": 5,"price": 180},

    {"id": 2003,"customer": "Chai", "city": "Bangkok","product":"Monitor",  "qty": 1,"price": 3200},

    {"id": 2004,"customer": "Duang", "city": "Phuket","product":"Mouse",    "qty": 3,"price": 180},

    {"id": 2005,"customer": "Anan", "city": "Korat","product":"Monitor",  "qty": 2,"price": 3200},

    {"id": 2006,"customer": "Ekk",  "city": "Bangkok","product":"Keyboard", "qty": 4, "price": 450},

]

tuple = []

for i in range(len(orders)):
    orders[i]["total"] = orders[i]["qty"] * orders[i]["price"] #จำนวนที่ซื้อ*ราคาต่อหน่วย
    orders[i]["vat"] = orders[i]["total"] * 0.07 #คำนวณ vat 7%
    orders[i]["total_vat"] = orders[i]["total"] + orders[i]["vat"] #ราคารวม vat สุทธิ
    tuple.append((orders[i]["id"],orders[i]["total"],orders[i]["total_vat"])) #เก็บข้อมูล รหัสคำสั่งซื้ (id) ราคารวม และราคารวม vat ลงใน list ของ tuple

print(tuple)
```
