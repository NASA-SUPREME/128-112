```python
#จงเขียนฟังก์ชัน safe_average(total, count) ทีคืนค่าเฉลียปัดทศนิยม 2 ตําแหน่ง และคืน None เมื่อ count ไม่เป็นบวก
def safe_average(total, count):
  if count <= 0:
    return None
  #คืนค่าเฉลียปัดทศนิยม 2 ตําแหน่ง
  return round(total/count,2)

#เขียนโปรแกรมหลักที่รับค่า PM2.5 จํานวน 5 วันด้วยตัวแปรห้าตัว แล้วใช้ตัวสะสมหาผลรวมและจํานวน
def main_program():
  #รับค่า PM2.5 จํานวน 5 วันด้วยตัวแปรห้าตัว
  PM1 = float(input("PM1: "))
  PM2 = float(input("PM2: "))
  PM3 = float(input("PM3: "))
  PM4 = float(input("PM4: "))
  PM5 = float(input("PM5: "))
  #ใช้ตัวสะสมหาผลรวมและจํานวน
  total = 0
  count = 0
  for PM in (PM1,PM2,PM3,PM4,PM5):
    if PM >=0:
      total += PM
      count += 1

  #ก่อนเรียก safe_average เพื่อแสดงค่าเฉลี่ย
  avg = safe_average(total,count)
  #โปรแกรมต้องทํางานถูกต้องด้วยเมื่อไม่มีข้อมูลทีใช้ได้เลย
  if avg is None:
    print(f"ไม่มีข้อมูลทีใช้ได้เลย")
  else:
    print(f"ค่าเฉลี่ย PM 2.5 = {avg} ({count} วัน)")

#เรียกใช้งาน
main_program()
```
